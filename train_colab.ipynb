## 3. Download & Extract Dataset from Hugging Face

Download  directly from your Hugging Face dataset repository into Colab.

```python
# Download dataset from Hugging Face
HF_REPO = "NikshanPs/inverse-fx-synthetic-dataset"

from huggingface_hub import hf_hub_download
import zipfile, os

print(f"Downloading dataset from Hugging Face: {HF_REPO}...")
zip_path = hf_hub_download(repo_id=HF_REPO, filename="synthetic_dataset.zip", repo_type="dataset")

print("Extracting dataset...")
with zipfile.ZipFile(zip_path, "r") as zip_ref:
    zip_ref.extractall("./")

# Verify files
!ls -lh dataset/synthetic/train/metadata.jsonl
!ls -lh dataset/synthetic/val/metadata.jsonl
print("Dataset ready for training!")
```

## 4. Train the Inverse FX Predictor Model on GPU
Trains **InverseFXNet** on the T4 GPU for 25 epochs.
*(Takes ~15-20 minutes on 4,000 samples)*

```python
!python3 train.py \
  --train_metadata dataset/synthetic/train/metadata.jsonl \
  --val_metadata dataset/synthetic/val/metadata.jsonl \
  --epochs 25 \
  --batch_size 32 \
  --lr 0.001 \
  --save_dir checkpoints \
  --device cuda
```

## 5. Evaluate Closed-Loop Performance with DAPN
Tests the trained model on an unseen validation track, predicts the audio FX chain, renders it through DAPN, and verifies the Multi-Resolution STFT reconstruction loss.

```python
!python3 evaluate_predictor.py \
  --checkpoint checkpoints/best_model.pth \
  --device cuda
```
